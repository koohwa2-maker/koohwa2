```python
#14.7 심사문제

k,e,m,s=map(int,input().split())
a=(k+e+m+s)/4
if a >= 80:
    print("합격")
    if a > 100:
        print("잘못된 점수")
else:
    print("불합격")
    
```

```python
k,e,m,s=map(int,input().split())
avg=(k+e+m+s)/4
if 0 <= k <= 100 and 0 <= e <= 100 and 0 <= m <= 100 and 0 <= s <= 100 :
    if avg >= 80:
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")
```
