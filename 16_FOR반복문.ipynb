```python
# for 와 Range 사용하기 

for i in range(100):
    print(" Hello, world?")
```

```python
for i in range(10):
    print(" Hello, world?",i+1)
```

```python
# 시작하는 숫자와 끝나는 숫자 지정하기

for b in range(6, 15):
    print(" hello",b)
```

```python
for c in range(6, 15, 2):
    print(" hello",c)
```

```python
# 숫자를 감소시커기 

for i in range(10,0,-1):
    print(i)
```

```python
a=[1,3,5,4,9]
list(reversed(a))
```

```python
for i in reversed(range(10)):
    print(i)
```
