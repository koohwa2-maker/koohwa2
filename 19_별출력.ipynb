```python
# 중첩루프 사용하기 

for i in range(4):
    for j in range(4):
        print("j:", j, sep="",end=" ")
    print("i:",i,"\\n",sep="")
```

```python
# 사각형으로 별출력하기 

for i in range(5):
    for j in range(5):
        print("*", end="")
    print()
```

```python
# 사각형 모양 바꾸기 

for i in range(3):
    for j in range(7):
        print("*", end="")
    print()
```

```python
# 계단으로 별 출력하기 

for i in range(5):
    for j in range(5):
        if j <= i :
            print("*", end="")
    print()
```

```python
 # 대각선으로 별 출력하기 

for i in range(5):
    for j in range(5):
        if j == i :
            print("*", end="")
        else:
            print(" ",end="")
    print()
```

```python
#19 연습문제

for i in range(5):
    for j in range(5):
        if j < i :
            print(" ", end="")
        else:
            print("*",end="")
    print()
```

```python
#19 심사문제: 산모양으로 별출력하기 

h=int(input())

for i in range(h):
    print(" "*(h-i-1),"*"*(2*i+1))
```
