```python
for i in range(1,31):
    if i % 3 ==0 and i % 5 ==0:
        print("FizzBuzz")
    elif i % 3 ==0:
            print("Fizz")      
    elif i % 5 ==0:
        print("Buzz")
    else:
        print(i)
```

```python
#20 심사문제 5와 7의 배수 , 공배수 처리하기

n,m = map(int,input().split())

if n > m:
    print("잘못된 범위입니다")
else:
    for i in range(n,m+1):
        if i % 5 == 0 and i % 7 == 0:
            print("FizzBuzz")
        elif i % 5 == 0:
                print("Fizz")      
        elif i % 7 == 0:
            print("Buzz")
        else:
            print(i)
```
