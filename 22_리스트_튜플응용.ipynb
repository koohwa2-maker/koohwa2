```python
# 인덱스와 요소를 ㅁ함께 출력하기

a=[38,21,53,62,19]
enumerate(a)
for index, value in enumerate(a):
    print(index, value)
```

```python
# while 반복문으로 요소 출력하기
a=[38,21,53,62,19]

i=0
while i < len(a):
    print(a[i])
    i += 1
```

```python
#가장 작은 수와 큰 수 구하기 
a=[38,21,53,62,19]

smallest = a[0]

for i in a:
    if i < smallest:
        smallest = i
print(smallest)
```

```python
#가장 작은 수와 큰 수 구하기 
a=[38,21,53,62,19]

largest = a[0]

for i in a:
    if i > largest:
        largest = i
print(largest)
```
