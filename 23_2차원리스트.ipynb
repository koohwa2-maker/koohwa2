```python
a=[
    [10,20],[30,40],[50,60]
]
for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j],end=" ")
    print()
```

```python
# while 반복문 한번 사용하기 

a=[
    [10,20],[30,40],[50,60]
]

i=0
while i < len(a):
    x,y =a[i]
    print(x,y)
    i +=1
```
